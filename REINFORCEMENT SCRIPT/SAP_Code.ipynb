```python
import os
path = r"C:\Users\ZB01760\Desktop\MIS-Data-Extraction-Script\REINFORCEMENT SCRIPT\MIS Reinforcements_Updated\Input_excels\SAP_All_In_One"
dir_list = os.listdir(path)
print("Files and directories in '", path, "' :")
print(dir_list)
```

# Code Logic

```python
# Code Logic by NVK

import re
import fsspec
import pandas as pd

final_ouput = pd.DataFrame()

for i in dir_list:
    if i.lower().startswith('fabric') == True:
        if re.search('squarewoven', i.lower()):
            path_new = path + '//' + str(i)

            fabric_sq_woven = pd.read_excel(path_new)

            fabric_sq_woven = fabric_sq_woven[['Material', 'Material description','Fabric Squarewoven Construction Code', 'Material type',
                                   'Fabric Processing Code', 'Warp Cord Densitiy', 'Cord Weight per 100 linear Yards',
                                   'Fabric Ends per Inch']]
            fabric_sq_woven = fabric_sq_woven.rename(columns={'Warp Cord Densitiy':'Cord Density Code', 
                                                            'Cord Weight per 100 linear Yards':'Linear Fill Cord Weight',
                                                            'Fabric Ends per Inch':'Calculated Ends per Inch'})

            fabric_sq_woven['Fabric Construction Code'] = ''
            fabric_sq_woven['Woven Fabric Width'] = ''
            fabric_sq_woven['Linear Fill Cord Weight DEN'] = ''
            fabric_sq_woven['Yarn Denier'] = ''
            fabric_sq_woven['Fill Cord'] = ''
            fabric_sq_woven['Picks per Meter'] = ''
            fabric_sq_woven['Wire Construction Code'] = ''
            fabric_sq_woven['Cable Construction Code'] = ''
            fabric_sq_woven = fabric_sq_woven[['Material', 'Yarn Denier', 'Material description', 'Material type',
                                'Fabric Construction Code', 'Fabric Processing Code',
                                'Cord Density Code', 'Fill Cord', 'Linear Fill Cord Weight',
                                'Picks per Meter','Woven Fabric Width', 'Calculated Ends per Inch', 'Linear Fill Cord Weight DEN',
                                'Fabric Squarewoven Construction Code','Wire Construction Code', 'Cable Construction Code']]
            fabric_sq_woven['Source'] = 'Fabrics Squarewoven all Fields'
            fabric_sq_woven['Cable Bead Coating'] = ''
            fabric_sq_woven['Gauge'] = ''
            fabric_sq_woven['Bead Inside Diameter'] = ''       
            fabric_sq_woven['Component Weight'] = ''   
            final_ouput = pd.concat([final_ouput, fabric_sq_woven], axis =0)
        else:
            print(i)
            paths_new = path + "//" + str(i)
            fab_all = pd.read_excel(paths_new)
            fabric = fab_all.copy()
            fabric_new =fabric[['Material', 'Yarn Denier', 'Material description', 'Material type',
                            'Fabric Construction Code', 'Fabric Processing Code.1',
                            'Cord Density Code', 'Fill Cord', 'Linear Fill Cord Weight',
                            'Picks per Meter','Woven Fabric Width', 'Calculated Ends per Inch', 'Linear Fill Cord Weight DEN']]
            if 'Fabric Processing Code' in fabric_new.columns:
                fabric_new = fabric_new.drop(columns=['Fabric Processing Code'])
            fabric_new = fabric_new.rename(columns= {'Fabric Processing Code.1':'Fabric Processing Code'})
            fabric_new['Fabric Squarewoven Construction Code'] = ''
            fabric_new['Wire Construction Code'] = ''
            fabric_new['Cable Construction Code'] = ''
            fabric_new['Source'] = 'Fabrics all Fields'
            fabric_new['Cable Bead Coating'] = ''
            fabric_new['Gauge'] = ''
            fabric_new['Bead Inside Diameter'] = ''       
            fabric_new['Component Weight'] = ''   
            fabric_new = fabric_new[fabric_new['Material'].str.startswith('RF')]
            final_ouput = pd.concat([final_ouput, fabric_new], axis =0)
    elif i.lower().startswith('single') ==True:

        paths = path + "//" + str(i)
        single_cord_fabric = pd.read_excel(paths)
        single_cord_fabric = single_cord_fabric[['Material', 'Material description', 'Material type', 'Fabric Construction Code',
                                                'Fabric Processing Code.1',  'Fill Cord',
                                                'Fabric Ends per Inch']]
        if 'Fabric Processing Code' in single_cord_fabric.columns:
            single_cord_fabric = single_cord_fabric.drop(columns=['Fabric Processing Code'])
            single_cord_fabric = single_cord_fabric.rename(columns={'Fabric Processing Code.1': 'Fabric Processing Code'})
        single_cord_fabric = single_cord_fabric.rename(columns={'Fabric Ends per Inch':'Calculated Ends per Inch'})

        single_cord_fabric['Woven Fabric Width'] = ''
        single_cord_fabric['Linear Fill Cord Weight DEN'] = ''
        single_cord_fabric['Yarn Denier'] = ''
        single_cord_fabric['Picks per Meter'] = ''
        single_cord_fabric['Cord Density Code'] = ''
        single_cord_fabric['Linear Fill Cord Weight'] = ''
        single_cord_fabric['Fabric Squarewoven Construction Code'] = ''
        single_cord_fabric['Wire Construction Code'] = ''
        single_cord_fabric['Cable Construction Code'] = ''
        single_cord_fabric = single_cord_fabric[['Material', 'Yarn Denier', 'Material description', 'Material type',
                            'Fabric Construction Code',
                            'Cord Density Code', 'Fill Cord', 'Linear Fill Cord Weight',
                            'Picks per Meter','Woven Fabric Width', 'Calculated Ends per Inch', 'Linear Fill Cord Weight DEN',
                            'Fabric Squarewoven Construction Code','Wire Construction Code', 'Cable Construction Code']]
        single_cord_fabric['Source'] = 'Single Cord Fabrics all Fields'
        single_cord_fabric['Cable Bead Coating'] = ''
        single_cord_fabric['Gauge'] = ''
        single_cord_fabric['Bead Inside Diameter'] = ''       
        single_cord_fabric['Component Weight'] = ''   
        single_cord_fabric = single_cord_fabric[single_cord_fabric['Material'].str.startswith('RF')]
        final_ouput = pd.concat([final_ouput, single_cord_fabric], axis =0)

    elif i.lower().startswith('wire') ==True:
        print("wires: ", i)
        paths = path + "//" + str(i)
        wire_all_fields= pd.read_excel(paths)
        wire_all_fields = wire_all_fields[['Material', 'Material description', 'Cable Construction Code', 'Wire Construction Code']]
        wire_all_fields['Fabric Construction Code'] = ''
        wire_all_fields['Fabric Processing Code'] = ''
        if 'Fabric Processing Code.1' in wire_all_fields.columns:
            wire_all_fields = wire_all_fields.drop(columns=['Fabric Processing Code'], errors='ignore')
            wire_all_fields = wire_all_fields.rename(columns={'Fabric Processing Code.1': 'Fabric Processing Code'})
        wire_all_fields['Cord Density Code'] = ''
        wire_all_fields['Fill Cord'] = ''
        wire_all_fields['Linear Fill Cord Weight'] = ''
        wire_all_fields['Picks per Meter'] = ''
        wire_all_fields['Woven Fabric Width'] = ''
        wire_all_fields['Calculated Ends per Inch'] = ''
        wire_all_fields['Linear Fill Cord Weight DEN'] = ''
        wire_all_fields['Yarn Denier'] = ''
        wire_all_fields['Material type'] = ''
        wire_all_fields['Fabric Squarewoven Construction Code'] = ''
        wire_all_fields = wire_all_fields[['Material', 'Yarn Denier', 'Material description', 'Material type',
                            'Fabric Construction Code', 'Fabric Processing Code',
                            'Cord Density Code', 'Fill Cord', 'Linear Fill Cord Weight',
                            'Picks per Meter','Woven Fabric Width', 'Calculated Ends per Inch', 'Linear Fill Cord Weight DEN', 
                            'Fabric Squarewoven Construction Code','Wire Construction Code', 'Cable Construction Code']]

        wire_all_fields['Source'] = 'Wires all Fields'
        wire_all_fields['Cable Bead Coating'] = ''
        wire_all_fields['Gauge'] = ''
        wire_all_fields['Bead Inside Diameter'] = ''       
        wire_all_fields['Component Weight'] = ''   
        wire_all_fields = wire_all_fields[wire_all_fields['Material'].str.startswith('RF')]
        final_ouput = pd.concat([final_ouput, wire_all_fields], axis =0)
    elif i.lower().startswith('cable') ==True:
        paths = path + "//" + str(i)
        cables = pd.read_excel(paths)
        cables_new = cables[['Material', 'Material description', 'Cable Bead Coating', 'Gauge',
                            'Bead Inside Diameter', 'Component Weight']]

        cables_new['Fabric Squarewoven Construction Code'] = ''
        cables_new['Wire Construction Code'] = ''
        cables_new['Cable Construction Code'] = ''
        cables_new['Source'] = 'Cable Beads'

        cables_new['Material type']=''
        cables_new['Fabric Construction Code']=''
        cables_new['Fabric Processing Code']=''
        if 'Fabric Processing Code.1' in cables_new.columns:
            cables_new = cables_new.drop(columns=['Fabric Processing Code'], errors='ignore')
            cables_new = cables_new.rename(columns={'Fabric Processing Code.1': 'Fabric Processing Code'})
        cables_new['Cord Density Code']=''
        cables_new['Fill Cord']=''
        cables_new['Linear Fill Cord Weight']=''
        cables_new['Picks per Meter']=''
        cables_new['Woven Fabric Width']=''
        cables_new['Calculated Ends per Inch']=''
        cables_new['Linear Fill Cord Weight DEN']=''
        cables_new['Yarn Denier']=''
        cables_new['Linear Fill Cord Weight']=''

        cables_new = cables_new[['Material', 'Yarn Denier', 'Material description', 'Material type',
                                'Fabric Construction Code', 'Fabric Processing Code',
                                'Cord Density Code', 'Fill Cord', 'Linear Fill Cord Weight',
                                'Picks per Meter','Woven Fabric Width', 'Calculated Ends per Inch', 'Linear Fill Cord Weight DEN', 
                                'Fabric Squarewoven Construction Code','Wire Construction Code', 'Cable Construction Code']]

        final_ouput = pd.concat([final_ouput, cables_new], axis =0)

final_ouput.reset_index(drop=True, inplace=True)      
```
